```sql
USE CATALOG online_retail;

CREATE SCHEMA IF NOT EXISTS gold;

-- ==========================================
-- GOLD TABLE 1: DAILY SALES SUMMARY
-- ==========================================

CREATE OR REPLACE TABLE gold.sales_summary AS

SELECT
    DATE(InvoiceDate) AS sales_date,
    COUNT(DISTINCT InvoiceNo) AS total_orders,
    SUM(Quantity) AS total_quantity_sold,
    ROUND(SUM(SalesAmount), 2) AS total_revenue,
    ROUND(
        SUM(SalesAmount) / COUNT(DISTINCT InvoiceNo),
        2
    ) AS average_transaction_value

FROM silver.online_retail_clean

GROUP BY DATE(InvoiceDate);


-- ==========================================
-- GOLD TABLE 2: COUNTRY SALES
-- ==========================================

CREATE OR REPLACE TABLE gold.country_sales AS

SELECT
    Country,
    COUNT(DISTINCT InvoiceNo) AS total_orders,
    SUM(Quantity) AS total_quantity_sold,
    ROUND(SUM(SalesAmount), 2) AS total_revenue

FROM silver.online_retail_clean

GROUP BY Country;


-- ==========================================
-- GOLD TABLE 3: PRODUCT SALES
-- ==========================================

CREATE OR REPLACE TABLE gold.product_sales AS

SELECT
    StockCode,
    Description,
    SUM(Quantity) AS total_quantity_sold,
    COUNT(DISTINCT InvoiceNo) AS total_orders,
    ROUND(SUM(SalesAmount), 2) AS total_revenue

FROM silver.online_retail_clean

GROUP BY StockCode, Description;
```

```sql
SELECT COUNT(*) AS sales_summary_rows
FROM online_retail.gold.sales_summary;

SELECT COUNT(*) AS country_sales_rows
FROM online_retail.gold.country_sales;

SELECT COUNT(*) AS product_sales_rows
FROM online_retail.gold.product_sales;
```

```sql
SELECT *
FROM online_retail.gold.sales_summary
ORDER BY sales_date
LIMIT 10;
```
