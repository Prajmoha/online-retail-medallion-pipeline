```sql
USE CATALOG online_retail;

CREATE SCHEMA IF NOT EXISTS silver;

CREATE OR REPLACE TABLE silver.online_retail_clean AS

SELECT
    TRIM(InvoiceNo) AS InvoiceNo,
    TRIM(StockCode) AS StockCode,
    TRIM(Description) AS Description,

    CAST(Quantity AS INT) AS Quantity,

    TO_TIMESTAMP(InvoiceDate, 'dd-MM-yyyy HH:mm') AS InvoiceDate,

    CAST(UnitPrice AS DECIMAL(10,2)) AS UnitPrice,

    CAST(CustomerID AS INT) AS CustomerID,

    TRIM(Country) AS Country,

    CAST(Quantity AS INT) *
    CAST(UnitPrice AS DECIMAL(10,2)) AS SalesAmount

FROM online_retail.bronze.online_retail_raw

WHERE
    InvoiceNo IS NOT NULL
    AND StockCode IS NOT NULL
    AND Quantity IS NOT NULL
    AND UnitPrice IS NOT NULL
    AND Quantity > 0
    AND UnitPrice > 0
    AND NOT STARTSWITH(TRIM(InvoiceNo), 'C');
```

```sql


SELECT COUNT(*) AS silver_row_count
FROM online_retail.silver.online_retail_clean;
```
